## 1. Iris
Load the data, drop flowers with sepal length > 5 cm, and plot petal vs. sepal length by species.

```python
# Load Libraries
from sklearn import datasets
import matplotlib.pyplot as plt
```

```python
iris_data = datasets.load_iris(as_frame=True)
iris = iris_data.frame
# Use each numeric label to look up its species name.
iris["species"] = iris_data.target_names[iris["target"]]

iris = iris[iris["sepal length (cm)"] <= 5]

iris.head()
```

Notes:
- An iris is a type of flower. 
- The sepal of a flower is the green, exterior part that surrounds the flower. 
- Sepal length is the length of the sepal (i.e. how long the green leaf-like part is). 
- Petal length is the length of the petal (i.e. the colorful, exterior part of the flower). 
- Measurements are in centimeters. 

Below, the scatter plot shows the relationship between the sepal length and the petal length for each species of iris covered in the data set. It is somewhat clear that species seem to be consistent in their petal length relative to their sepal length with clear lines for division, e.g., via using a decision tree.

```python
species_colors = {
    "setosa": "red",
    "versicolor": "green",
    "virginica": "blue"
}

for species, color in species_colors.items():
    subset = iris[iris["species"] == species]
    plt.scatter(subset["sepal length (cm)"], subset["petal length (cm)"], 
                label=species, color=color)

plt.xlabel("Sepal length (cm)")
plt.ylabel("Petal length (cm)")
plt.legend()
plt.show()
```

## 2. Handwritten digits
Keep only 3s and 8s and display examples. This uses the 8×8 digits dataset from [here](https://scikit-learn.org/stable/auto_examples/classification/plot_digits_classification.html) with the native loader from sklearn.

```python
# Load the digits data set into a pandas data frame
digits = datasets.load_digits(as_frame=True).frame

# Keep only 3s and 8s
digits = digits[digits["target"].isin([3, 8])]
```

```python
# Visualize the first 20 digits
images = digits.drop(columns="target").to_numpy().reshape(-1, 8, 8)

plt.figure(figsize=(12, 3))
for i in range(20):
    plt.subplot(2, 10, i + 1)
    plt.imshow(images[i], cmap="gray_r")
    plt.axis("off")
plt.show()
```

Notes:
- Each image is a hand-written digit. They seem pixelated and have variation in how the number is written as you'd expect. 
- Each row contains 64 pixel values (each pixel is a value between 0 and 16, signifying the darkness of the pixel) and a label for what digit it corresponds to.
- The first 20 are reshaped into 8×8 images and displayed in a grid.
